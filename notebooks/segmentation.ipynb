# Semantic segmentation
In this notebook you will learn how to do semantic segmentation with the VisionPipelines library.

```python
# Set up modules directory to import the library
import sys
import os
sys.path.append(os.path.abspath(os.path.join('..', 'src')))
```

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np
from visionpipelines.constants import SegmentationMethod
from visionpipelines.pipelines.segmentation_pipeline import SegmentationPipeline
```

```python
# read test image
image_path1 = '../tests/data/people_image.webp'
image1 = cv2.imread(image_path1)
```

```python
imgplot1 = plt.imshow(cv2.cvtColor(image1, cv2.COLOR_BGR2RGB))
```

```python
# Use DeepLabV3 for semantic segmentation
pipeline = SegmentationPipeline(SegmentationMethod.DEEPLABV3)

# Perform segmentation
mask = pipeline.run_pipeline(image1)

image_with_mask = pipeline.overlay_mask(image1, mask)
```

```python
# display segmentation mask overlaid on the original image
plt.imshow(cv2.cvtColor(image_with_mask, cv2.COLOR_BGR2RGB))
```

```python
# classes detected in the image
detected_classes = [pipeline.segmenter.categories[i] for i in np.unique(mask)]
detected_classes
```
